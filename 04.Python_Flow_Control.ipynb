# 2.Loop Statement 
A Loop is a programing comstruct that executes a block of code repeatedly, as ling as a speified condition is satisfied or until all items sequece have been processed 

In Hindi : Jab hemen ek hi code ko bar bar execute krna hota he tab hum loop ka use karte hain.

```python
for i in range(5,51):
        if i%5==0:
            print(i)          
```

```python
for i in range(5):
    print('Hello,World!')
```

# For Loop :~
A for loop is used to iterate over the items of a sequence or another iterable object, executing a block of code for each item. 

For loop ka use tab karte he jab hame kisi sequence ka elements pae ek ek karke operation perform krte he .

```python
for i in range(10):
    print(i)
```

```python
a=int(input("enter your number"))
for i in range(1,10):
    print(a ,'x', i ,'=', i*a)
```

```python
#Range(): range() ka use integer sequence represent karne ke liye bhi hota he, jo for loop me bhu commanly use kiya jata he.
```

```python
# Forms in Range :~
# 1. Range(stop)
for i in range(5):
    print(i)
```

```python
# 2. Range(start,stop)
for i in range(10,20):
    print(i)
```

```python
# 3. Range(start,stop,step,)
for i in range(2,11,2):
    print(i)
```

```python
# negative
for i in range(10,0,-1):
    print(i)                
```

```python
# For loop with different data types :~ 
fruit=['Apple','Banana','Mango','Orange']
for i in fruit:
    print(i)
```

```python
# Loop wiht String:~
name='Tushar'
for i in name:
    print(i)
```

```python
vowels='aeiou'
for i in vowels:
    print(i)
```
