# Indexing

```python
# Print the second item of the list 
course=['Data Science','Machine Learning','Python','SQL','Linux']
print(course)
print(course[1])
```

```python
# Positive Indexing :
print(course[0])
print(course[3])
```

```python
# Negative indexing :
course=['Data Science','Machine Learning','Python','SQL','Linux'] 
print(course[-1])
print(course[-4])
```

```python
# Length of list 
len(course)
```
