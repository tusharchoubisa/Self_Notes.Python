# Data Type 

```python
# 1 Strings 
# all the name in words 
a = 'Tushar'
b = 'a'
print(a)
print(b)
```

```python
#2 Float :
b= 90.50
print(b)
type(b)
```

```python
#3 complex :
a = b+9j
print(a)
type(a)
```

```python
#4 Boolean :
a = True 
b = False 
print(a)
print(b)
type(a)
type(b)
```

```python
#5 None :
a = None 
print(a)
type(a)
```

```python
#6 Collection data type :

# i) List :
a = [ 1,2,3,4,5,6,7,8]
print(a)
type(a)
```

```python
#List under list :
a = [[ 1,2,3,4,5,6,7,8],['a','b','c','d','e','f']]
print(a)
```

```python
#ii) TUPLE :
a=(1,2,3,4,5,5,6,7,8)
print(a)
type(a)
```

```python
# iii) Dictionary 
students = {
      'Name' : 'tushar',
      'city ' : 'udaipur',
      'age' :  21 ,
    'Marks' : 56 
}
print(students)
print(type(students))
```

```python
# iv) Set :
a={1,2,3,4,5,5,6,7,8}
print(a)
```

```python
# All Data Types ;
a =23  #integer
print(type(a),a)
b = 54.5 #float 
print(type(b),b)
c = [1,2,3,4,5,'rahul','tushar',8] #list 
print(type(c),c)
d = (1,2,34,5,6,7,5,6) #tuple 
print(type(d),d)
e = { 
    'nam' : 'tushar' , #dictionary 
    'age' : 22 
} 
print(type(e),e)
f= {1,2,3,4,6,5}   #set 
print(type(f),f)
```

# Type Converstion 

```python
# Type Converstion :
#1) Implicit : What input we have that output will Come like int,float,string .
a = 10  #float
b = 50.70
print(a+b)
a = 50  #int
b = 50 
print(a+b)

#2 Explicit : We make the output data type like we have a input int we want a
#float output we can make .
a = 50
b = 40 
print(a+float(b))

a= 50
b = 40.50
print(a+int(b))
```
