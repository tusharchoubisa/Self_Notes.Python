# List Method 

# 1. append() 
The append() method is adds single elements to the end of an existing list.

```python
a=[10,20,30,40,50,60,70]
a.append(80)
a
```

```python
fruits=['apple','banana','mango']
fruits.append('chiku')
fruits
```

# 2. Insert()
The insert method adds an element at a specified Position in a list.

```python
fruits=['apple','banana','mango']
fruits.insert(2,'orange')
fruits
```

```python
a=[10,20,30,40,50,60,70]
a.insert(3,34)
a
```

# 3. Extend()
The Extend method add all elements from an iterable to the end of an existing list.

```python
list1=[10,20,30,40,50,60]
list2=[2,3,4,5,6,7,8,9]
list1.extend(list2)
list1
```

```python
fruits=['apple','banana','mango']
course=['Data science','machine learning','pyhton','ststicitic','Big data']
fruits.extend(course)
print(fruits)
```

# 4. Sort()

~The sort() method arrenges the elements of a list in ascending order ny default.

```python
num=[90,40,50,60,32,10,30,20]
num.sort()
num
```

```python
num=[90,40,50,60,32,10,30,20]
num.sort(reverse=True)
num
```

```python
num=[90,40,50,60,32,10,30,20]
num.sort(reverse=False)
num
```

```python
names=['tushar','rahu','shreya','harsh']
names.sort()
names
```
